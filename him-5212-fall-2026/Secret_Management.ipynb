```python
# CMS Blue Button API credentials
# NOTE: After running this notebook to store the secret, clear the value below
# and retrieve it from the secret scope using dbutils.secrets.get() instead.
# client_id is a public identifier and is NOT stored in the secret scope.

SCOPE_NAME = "cms_blue_button_api"

client_secret = ""
```

```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()

# Create the secret scope (will raise if it already exists — that's OK)
try:
    w.secrets.create_scope(scope=SCOPE_NAME)
    print(f"Secret scope '{SCOPE_NAME}' created.")
except Exception as e:
    if "already exists" in str(e).lower() or "RESOURCE_ALREADY_EXISTS" in str(e):
        print(f"Secret scope '{SCOPE_NAME}' already exists — continuing.")
    else:
        raise

# Store the client_secret in the scope (client_id is public — not stored)
w.secrets.put_secret(scope=SCOPE_NAME, key="client_secret", string_value=client_secret)

print(f"Secret 'client_secret' stored in scope '{SCOPE_NAME}'.")
```

```python
# List secrets in the scope (values are never returned by design)
secrets = w.secrets.list_secrets(scope=SCOPE_NAME)
print(f"Secrets in scope '{SCOPE_NAME}':")
for s in secrets:
    print(f"  - {s.key}")

# Verify retrieval using dbutils (this is how other notebooks will access the secret)
retrieved_secret = dbutils.secrets.get(scope=SCOPE_NAME, key="client_secret")
print(f"\nclient_secret retrieved: {'success' if retrieved_secret else 'failed'}")
print(f"\nUsage in other notebooks:")
print(f"  dbutils.secrets.get(scope='{SCOPE_NAME}', key='client_secret')")
```
